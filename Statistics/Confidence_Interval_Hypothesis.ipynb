```python
import sys
import numpy as np
from scipy import stats

print("Python:", sys.version.split()[0])
print("Interpreter:", sys.executable)
print("NumPy:", np.__version__)
print("SciPy:", stats.__version__ if hasattr(stats, "__version__") else "OK (stats loaded)")
```

```python
data = np.array([52, 49, 51, 50, 53, 47, 54, 52, 49, 51])
n = len(data)
xbar = data.mean()
s = data.std(ddof=1)              # sample std uses n-1
se = s / np.sqrt(n)               # standard error
alpha = 0.05
df = n - 1

t_crit = stats.t.ppf(1 - alpha/2, df)
ci = (xbar - t_crit * se, xbar + t_crit * se)

print("n:", n)
print("x̄:", xbar)
print("s:", s)
print("SE:", se)
print("t_crit:", t_crit)
print("95% CI for μ:", ci)
```

```python
mu0 = 50
t_stat, p_value = stats.ttest_1samp(data, popmean=mu0, alternative="two-sided")

print("t-stat:", t_stat)
print("p-value:", p_value)

alpha = 0.05
print("Decision:", "Reject H0" if p_value < alpha else "Fail to reject H0")
```

```python
print("Is μ0 inside the 95% CI?", ci[0] <= mu0 <= ci[1])
```

```python
x = 62
n = 100
phat = x / n
alpha = 0.05

z_crit = stats.norm.ppf(1 - alpha/2)
se_p = np.sqrt(phat * (1 - phat) / n)
ci_p = (phat - z_crit * se_p, phat + z_crit * se_p)

print("p̂:", phat)
print("SE(p̂):", se_p)
print("95% CI for p:", ci_p)
```

```python
p0 = 0.50
phat = x / n

se0 = np.sqrt(p0 * (1 - p0) / n)      # SE under H0 uses p0
z = (phat - p0) / se0
p_value = 2 * (1 - stats.norm.cdf(abs(z)))  # two-tailed

print("z:", z)
print("p-value:", p_value)
print("Decision:", "Reject H0" if p_value < 0.05 else "Fail to reject H0")
```
