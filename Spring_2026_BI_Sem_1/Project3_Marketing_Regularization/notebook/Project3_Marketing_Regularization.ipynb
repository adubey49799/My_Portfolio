```python
from pathlib import Path
import pandas as pd

DATA_PATH = Path("../data")  # because notebook is inside notebook/ folder
print("Data folder exists:", DATA_PATH.exists())
print("Files in data folder:")
for f in DATA_PATH.glob("*"):
    print(" -", f.name)
```

```python
csv_path = Path("../data/marketing_ab.csv")  # expected name

if not csv_path.exists():
    raise FileNotFoundError(
        f"Could not find {csv_path}. Rename your Kaggle CSV to marketing_ab.csv "
        "or update this path to match the actual file name you saw above."
    )

df = pd.read_csv(csv_path)

print("Shape:", df.shape)
print("Columns:", list(df.columns))
df.head()
```

```python
print(df["converted"].value_counts())
print("Conversion rate:", df["converted"].mean())
```

```python
# Drop unnecessary columns
df = df.drop(columns=["Unnamed: 0", "user id"])

# Rename columns for clean modeling
df = df.rename(columns={
    "test group": "test_group",
    "total ads": "total_ads",
    "most ads day": "most_ads_day",
    "most ads hour": "most_ads_hour"
})

print("Cleaned Columns:")
print(df.columns)
df.head()
```

```python
X = df[["test_group", "total_ads", "most_ads_day", "most_ads_hour"]]
y = df["converted"].astype(int)  # convert True/False → 1/0

print("X shape:", X.shape)
print("y mean (conversion rate):", y.mean())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)
print("Train conversion rate:", y_train.mean())
print("Test conversion rate:", y_test.mean())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

categorical_features = ["test_group", "most_ads_day"]
numeric_features = ["total_ads", "most_ads_hour"]

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(drop="first", handle_unknown="ignore"))
])

preprocess = ColumnTransformer(
    transformers=[
        ("cat", categorical_transformer, categorical_features),
        ("num", numeric_transformer, numeric_features),
    ]
)

preprocess
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

baseline_model = Pipeline(steps=[
    ("preprocess", preprocess),
    ("model", LogisticRegression(penalty=None, max_iter=5000))
])

baseline_model.fit(X_train, y_train)

print("Baseline Model Trained")
```

```python
from sklearn.metrics import roc_auc_score, average_precision_score
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

y_pred_proba = baseline_model.predict_proba(X_test)[:, 1]
y_pred = baseline_model.predict(X_test)

print("=== Baseline Performance ===")
print("ROC-AUC:", roc_auc_score(y_test, y_pred_proba))
print("PR-AUC:", average_precision_score(y_test, y_pred_proba))
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1:", f1_score(y_test, y_pred))
```

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold

param_grid = {
    "model__C": [0.001, 0.01, 0.1, 1, 10]
}

ridge_model = Pipeline(steps=[
    ("preprocess", preprocess),
    ("model", LogisticRegression(
        penalty="l2",
        solver="lbfgs",
        max_iter=5000
    ))
])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

ridge_grid = GridSearchCV(
    ridge_model,
    param_grid,
    cv=cv,
    scoring="roc_auc",
    n_jobs=-1
)

ridge_grid.fit(X_train, y_train)

print("Best C (Ridge):", ridge_grid.best_params_)
print("Best CV ROC-AUC:", ridge_grid.best_score_)
```

```python
ridge_best = ridge_grid.best_estimator_

y_pred_proba = ridge_best.predict_proba(X_test)[:, 1]
y_pred = ridge_best.predict(X_test)

print("=== Ridge Performance ===")
print("ROC-AUC:", roc_auc_score(y_test, y_pred_proba))
print("PR-AUC:", average_precision_score(y_test, y_pred_proba))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1:", f1_score(y_test, y_pred))
```

```python
lasso_model = Pipeline(steps=[
    ("preprocess", preprocess),
    ("model", LogisticRegression(
        penalty="l1",
        solver="saga",
        max_iter=5000
    ))
])

lasso_grid = GridSearchCV(
    lasso_model,
    param_grid,
    cv=cv,
    scoring="roc_auc",
    n_jobs=-1
)

lasso_grid.fit(X_train, y_train)

print("Best C (Lasso):", lasso_grid.best_params_)
print("Best CV ROC-AUC:", lasso_grid.best_score_)
```

```python
lasso_best = lasso_grid.best_estimator_

y_pred_proba = lasso_best.predict_proba(X_test)[:, 1]
y_pred = lasso_best.predict(X_test)

print("=== Lasso Performance (Test Set) ===")
print("ROC-AUC:", roc_auc_score(y_test, y_pred_proba))
print("PR-AUC:", average_precision_score(y_test, y_pred_proba))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1:", f1_score(y_test, y_pred))
```

```python
import numpy as np

def get_feature_names(preprocessor):
    cat_features = preprocessor.named_transformers_["cat"]\
        .named_steps["onehot"]\
        .get_feature_names_out(["test_group", "most_ads_day"])

    num_features = ["total_ads", "most_ads_hour"]

    return list(cat_features) + num_features

feature_names = get_feature_names(lasso_best.named_steps["preprocess"])
feature_names[:10]
```

```python
coefficients = lasso_best.named_steps["model"].coef_.flatten()

coef_df = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients
})

coef_df["Abs_Coefficient"] = coef_df["Coefficient"].abs()
coef_df = coef_df.sort_values("Abs_Coefficient", ascending=False)

coef_df.head(15)
```

```python
zero_features = coef_df[coef_df["Coefficient"] == 0]["Feature"].tolist()

print("Number of features removed by Lasso:", len(zero_features))
print("Removed Features:")
zero_features
```

```python
coef_df_nonzero = coef_df[coef_df["Coefficient"] != 0].copy()
coef_df_nonzero["Odds_Ratio"] = np.exp(coef_df_nonzero["Coefficient"])

coef_df_nonzero
```
