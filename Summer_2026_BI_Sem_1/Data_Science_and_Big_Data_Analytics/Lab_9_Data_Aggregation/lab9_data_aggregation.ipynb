## Step 1: Dataset Selection

For this lab, I selected the Red Wine Quality dataset. This dataset contains physicochemical properties of red wine samples, such as fixed acidity, volatile acidity, citric acid, residual sugar, chlorides, density, pH, sulphates, alcohol, and quality score.

This dataset is useful for data aggregation because most of the columns are numeric. The `quality` column can also be used as a grouping variable to compare average wine characteristics across different quality ratings.

Dataset source: Public Red Wine Quality dataset from Kaggle/UCI Machine Learning Repository.

```python
# Lab 9: Data Aggregation
# Import required libraries

import pandas as pd
import numpy as np

print("Libraries imported successfully.")
```

## Step 2: Load the Dataset

In this step, the Red Wine Quality dataset is loaded into a Pandas DataFrame. After loading the data, the first 10 rows are displayed to understand the structure of the dataset.

```python
# Load the dataset into a Pandas DataFrame

df = pd.read_csv("winequality-red.csv")

print("Dataset loaded successfully.")
```

```python
# Display the first 10 rows of the dataset

df.head(10)
```

## Step 3: Explore the Dataset Structure

Before performing aggregation, it is important to understand the structure of the dataset. In this step, I checked the number of rows and columns, reviewed the column names, and inspected the data types.

```python
# Display the shape of the dataset

print("Dataset shape:")
print(df.shape)
```

```python
# Display column names

print("Column names:")
print(df.columns.tolist())
```

```python
# Display data types

print("Data types:")
print(df.dtypes)
```

## Step 4: Select Numeric Columns

The aggregation functions are applied to numeric columns in the dataset. Since the Red Wine Quality dataset mostly contains numeric values, these columns are suitable for summary statistics such as mean, median, minimum, maximum, standard deviation, variance, and sum.

```python
# Select numeric columns from the dataset

numeric_df = df.select_dtypes(include=["number"])

print("Numeric columns selected for aggregation:")
print(numeric_df.columns.tolist())
```

```python
# Display first 5 rows of numeric columns

numeric_df.head()
```

## Step 5: Apply Built-in Pandas Aggregation Functions

In this step, I applied the required built-in Pandas aggregation functions to the numeric columns. These functions help summarize the dataset by showing counts, first values, last values, averages, medians, minimums, maximums, standard deviations, variances, and totals.

```python
# Count non-null values in each numeric column

count_result = numeric_df.count()

print("Count of non-null values:")
print(count_result)
```

```python
# Display first value from each numeric column

first_result = numeric_df.iloc[0]

print("First value in each numeric column:")
print(first_result)
```

```python
# Display last value from each numeric column

last_result = numeric_df.iloc[-1]

print("Last value in each numeric column:")
print(last_result)
```

```python
# Calculate mean of each numeric column

mean_result = numeric_df.mean()

print("Mean of numeric columns:")
print(mean_result)
```

```python
# Calculate median of each numeric column

median_result = numeric_df.median()

print("Median of numeric columns:")
print(median_result)
```

```python
# Calculate minimum value of each numeric column

min_result = numeric_df.min()

print("Minimum value of numeric columns:")
print(min_result)
```

```python
# Calculate maximum value of each numeric column

max_result = numeric_df.max()

print("Maximum value of numeric columns:")
print(max_result)
```

```python
# Calculate standard deviation of each numeric column

std_result = numeric_df.std()

print("Standard deviation of numeric columns:")
print(std_result)
```

```python
# Calculate variance of each numeric column

var_result = numeric_df.var()

print("Variance of numeric columns:")
print(var_result)
```

```python
# Calculate sum of each numeric column

sum_result = numeric_df.sum()

print("Sum of numeric columns:")
print(sum_result)
```

## Step 6: Create a Combined Aggregation Summary Table

Instead of viewing each aggregation result separately, I created one combined summary table using the Pandas `agg()` function. This table provides a compact view of the major summary statistics for all numeric columns.

```python
# Create a combined aggregation summary table

aggregation_summary = numeric_df.agg([
    "count",
    "mean",
    "median",
    "min",
    "max",
    "std",
    "var",
    "sum"
])

aggregation_summary
```

# Save the aggregation summary table as a CSV file

aggregation_summary.to_csv("lab9_aggregation_summary.csv")

print("Aggregation summary saved as lab9_aggregation_summary.csv")

```python
# Save the aggregation summary table as a CSV file

aggregation_summary.to_csv("lab9_aggregation_summary.csv")

print("Aggregation summary saved as lab9_aggregation_summary.csv")
```

## Step 7: Groupby Analysis

In this step, I used the `groupby()` function to group the dataset by the `quality` column. The `quality` column represents the wine quality score. By grouping the data by quality and calculating the mean of the numeric columns, I can compare how the average chemical properties differ across wine quality levels.

```python
# Group the dataset by wine quality and calculate mean values

groupby_quality_mean = df.groupby("quality").mean(numeric_only=True)

groupby_quality_mean
```

```python
# Save the groupby quality summary as a CSV file

groupby_quality_mean.to_csv("lab9_groupby_quality_mean.csv")

print("Groupby quality mean summary saved as lab9_groupby_quality_mean.csv")
```

```python
# Count number of wine samples for each quality score

quality_counts = df.groupby("quality")["quality"].count()

quality_counts
```

```python
# Save quality counts as a CSV file

quality_counts.to_csv("lab9_quality_counts.csv")

print("Quality counts saved as lab9_quality_counts.csv")
```

## Step 8: Observations from Aggregation and Groupby Analysis

After applying aggregation functions and performing the groupby analysis, several observations can be made from the Red Wine Quality dataset.

```python
# Display selected columns from the groupby result for easier interpretation

selected_groupby_columns = groupby_quality_mean[
    ["volatile acidity", "citric acid", "sulphates", "alcohol", "pH"]
]

selected_groupby_columns
```

```python
# Identify the average alcohol level by wine quality

alcohol_by_quality = df.groupby("quality")["alcohol"].mean()

print("Average alcohol level by wine quality:")
print(alcohol_by_quality)
```

```python
# Identify the average volatile acidity by wine quality

volatile_acidity_by_quality = df.groupby("quality")["volatile acidity"].mean()

print("Average volatile acidity by wine quality:")
print(volatile_acidity_by_quality)
```

```python
# Identify the average sulphates level by wine quality

sulphates_by_quality = df.groupby("quality")["sulphates"].mean()

print("Average sulphates level by wine quality:")
print(sulphates_by_quality)
```

### Observations

The aggregation functions provided a clear statistical summary of the numeric columns in the dataset. The `count()` function showed that each numeric column had the same number of non-null records, which indicates that the dataset did not contain missing values in these columns. The `mean()` and `median()` values helped describe the central tendency of each variable, while `min()` and `max()` showed the range of values. The `std()` and `var()` functions showed how much variation existed in each column.

The groupby analysis showed that wine characteristics vary across quality scores. In particular, alcohol content appears to be an important feature because the average alcohol level generally increases as wine quality increases. Volatile acidity appears to move in the opposite direction, with lower average volatile acidity often associated with higher wine quality. Sulphates also show variation across quality groups and may help explain differences in wine quality.

Overall, the groupby analysis helped compare average chemical properties across wine quality ratings and made it easier to identify patterns that would not be obvious by looking only at individual rows.
