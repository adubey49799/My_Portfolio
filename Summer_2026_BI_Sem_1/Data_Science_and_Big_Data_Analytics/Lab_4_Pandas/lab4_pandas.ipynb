## Setup

```python
# Import pandas library
import pandas as pd
import numpy as np
from pathlib import Path

# Display more columns and wider output for easier review
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

# Define the input Excel file path
excel_file = Path("Lab 4 - Automobile.xls")

# Define the cleaned CSV output file path
cleaned_csv_file = Path("Lab_4_Automobile_Cleaned.csv")

print("Pandas version:", pd.__version__)
print("Current working directory:", Path.cwd())
print("Expected Excel file:", excel_file)
```

# Part 2: Working with the Automobile Dataset

## Task 3: Load the dataset using Pandas

```python
# Load the Excel file into a Pandas DataFrame
# xlrd is required for reading .xls files.
auto_df = pd.read_excel(excel_file)

# Display basic dataset information
print("Task 3: Automobile dataset loaded successfully.")
print("Shape of dataset:", auto_df.shape)
print("\nColumn names:")
print(auto_df.columns.tolist())
```

## Task 4: Print the first five and last five rows

```python
# Display first five rows
print("First five rows:")
display(auto_df.head())

# Display last five rows
print("Last five rows:")
display(auto_df.tail())
```

## Task 5: Remove records containing '?', 'n.a', or NaN values and save cleaned DataFrame

```python
# Replace invalid values '?' and 'n.a' with NaN
auto_cleaned_df = auto_df.replace(["?", "n.a"], np.nan)

# Remove rows containing any NaN values
auto_cleaned_df = auto_cleaned_df.dropna()

# Save cleaned DataFrame to CSV
auto_cleaned_df.to_csv(cleaned_csv_file, index=False)

# Display result
print("Original dataset shape:", auto_df.shape)
print("Cleaned dataset shape:", auto_cleaned_df.shape)
print("Cleaned CSV saved as:", cleaned_csv_file)

display(auto_cleaned_df.head())
```

## Helper Cell: Identify important columns

Different Automobile datasets may use slightly different column names.  
This cell detects likely company, price, and mileage columns.

```python
# Helper function to find a column by possible names
def find_column(df, possible_names):
    normalized_columns = {col.lower().strip().replace(" ", "_").replace("-", "_"): col for col in df.columns}
    for name in possible_names:
        key = name.lower().strip().replace(" ", "_").replace("-", "_")
        if key in normalized_columns:
            return normalized_columns[key]
    return None

# Detect commonly used columns
company_col = find_column(auto_cleaned_df, ["company", "make", "manufacturer", "car_company"])
price_col = find_column(auto_cleaned_df, ["price"])
mileage_col = find_column(auto_cleaned_df, ["average_mileage", "average-mileage", "mileage", "city_mpg", "highway_mpg"])

print("Detected company column:", company_col)
print("Detected price column:", price_col)
print("Detected mileage column:", mileage_col)

# Convert price and mileage columns to numeric if they exist
if price_col:
    auto_cleaned_df[price_col] = pd.to_numeric(auto_cleaned_df[price_col], errors="coerce")

if mileage_col:
    auto_cleaned_df[mileage_col] = pd.to_numeric(auto_cleaned_df[mileage_col], errors="coerce")
```

## Task 6: Find and print the most expensive car

```python
# Find the most expensive car based on the price column
most_expensive_car = auto_cleaned_df.loc[auto_cleaned_df[price_col].idxmax()]

print("Most expensive car in the dataset")
print("Company:", most_expensive_car[company_col])
print("Price:", most_expensive_car[price_col])
print("\nFull details:")
display(most_expensive_car.to_frame().T)
```

## Task 7: Print all details of cars manufactured by Toyota

```python
# Filter cars manufactured by Toyota
toyota_cars = auto_cleaned_df[auto_cleaned_df[company_col].astype(str).str.lower() == "toyota"]

print("Toyota cars:")
display(toyota_cars)
```

## Task 8: Count and display total number of cars per company

```python
# Count number of cars per company
cars_per_company = auto_cleaned_df[company_col].value_counts()

print("Total number of cars per company:")
print(cars_per_company)
```

## Task 9: Display the most expensive car from each company

```python
# Sort by company and price, then keep the highest priced car for each company
most_expensive_by_company = (
    auto_cleaned_df.sort_values(by=[company_col, price_col], ascending=[True, False])
    .groupby(company_col)
    .head(1)
    .reset_index(drop=True)
)

print("Most expensive car from each company:")
display(most_expensive_by_company)
```

## Task 10: Calculate average mileage for each car company

```python
# Calculate average mileage by company
average_mileage_by_company = (
    auto_cleaned_df.groupby(company_col)[mileage_col]
    .mean()
    .sort_values(ascending=False)
)

print("Average mileage by company:")
print(average_mileage_by_company)
```

## Task 11: Sort all cars in ascending order based on price

```python
# Sort all cars by price in ascending order
cars_sorted_by_price = auto_cleaned_df.sort_values(by=price_col, ascending=True)

print("Cars sorted by price in ascending order:")
display(cars_sorted_by_price)
```

# Part 3: Combining DataFrames

## Task 12: Create two DataFrames and concatenate them

```python
# Create dictionaries for German and Japanese cars
GermanCars = {
    "Company": ["Ford", "Mercedes", "BMV", "Audi"],
    "Price": [23845, 171995, 135925, 71400]
}

japaneseCars = {
    "Company": ["Toyota", "Honda", "Nissan", "Mitsubishi "],
    "Price": [29995, 23600, 61500, 58900]
}

# Convert dictionaries to DataFrames
german_cars_df = pd.DataFrame(GermanCars)
japanese_cars_df = pd.DataFrame(japaneseCars)

# Concatenate both DataFrames
combined_cars_df = pd.concat([german_cars_df, japanese_cars_df], ignore_index=True)

print("German Cars DataFrame:")
display(german_cars_df)

print("Japanese Cars DataFrame:")
display(japanese_cars_df)

print("Concatenated Cars DataFrame:")
display(combined_cars_df)
```

## Task 13: Merge Car_Price and Car_Horsepower DataFrames

```python
# Create dictionaries for car price and horsepower
Car_Price = {
    "Company": ["Toyota", "Honda", "BMV", "Audi"],
    "Price": [23845, 17995, 135925, 71400]
}

car_Horsepower = {
    "Company": ["Toyota", "Honda", "BMV", "Audi"],
    "horsepower": [141, 80, 182, 160]
}

# Convert dictionaries to DataFrames
car_price_df = pd.DataFrame(Car_Price)
car_horsepower_df = pd.DataFrame(car_Horsepower)

# Merge DataFrames on Company column
merged_car_df = pd.merge(car_price_df, car_horsepower_df, on="Company")

print("Car Price DataFrame:")
display(car_price_df)

print("Car Horsepower DataFrame:")
display(car_horsepower_df)

print("Merged DataFrame:")
display(merged_car_df)
```
