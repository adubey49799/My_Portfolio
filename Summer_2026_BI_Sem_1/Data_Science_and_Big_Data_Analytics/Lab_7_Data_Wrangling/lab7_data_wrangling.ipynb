```python
# Lab 7: Data Wrangling
# Step 1: Import required libraries and read the dataset

import pandas as pd
import numpy as np

# Read the Excel dataset
# Make sure this file is in the same folder as your notebook
cancer_df = pd.read_excel("Lab 7 - Cancer.xls")

# Display the first 5 rows
cancer_df.head()
```

```python
# Step 2: Display the data types of each column

cancer_df.dtypes
```

```python
# Step 3: Generate summary statistics for numeric columns

cancer_df.describe()
```

```python
# Step 4: Group the dataset by class and doctor_name

grouped_df = cancer_df.groupby(['class', 'doctor_name']).size().reset_index(name='record_count')

# Display grouped result
grouped_df
```

```python
# Step 5: Provide a summary of missing values within the dataset

missing_values_summary = cancer_df.isnull().sum()

# Display missing values column-wise
missing_values_summary
```

```python
# Step 6: Drop all rows that contain missing data

cancer_df_clean = cancer_df.dropna()

# Display the original and cleaned dataset shapes
print("Original dataset shape:", cancer_df.shape)
print("Dataset shape after dropping missing values:", cancer_df_clean.shape)

# Display first 5 rows of cleaned dataset
cancer_df_clean.head()
```

```python
# Step 7: Summarize the number of unique values in each column

unique_values_summary = cancer_df_clean.nunique()

# Display number of unique values column-wise
unique_values_summary
```

```python
# Step 8: Display unique value details for each column

for column in cancer_df_clean.columns:
    print(f"\nColumn: {column}")
    print(f"Number of unique values: {cancer_df_clean[column].nunique()}")
    print("Sample unique values:")
    print(cancer_df_clean[column].unique()[:10])
```

```python
# Step 9: Find duplicate values in patient_id

patient_id_counts = cancer_df_clean['patient_id'].value_counts()

# Display patient_id values that appear more than once
duplicate_patient_ids = patient_id_counts[patient_id_counts > 1]

duplicate_patient_ids
```

```python
# Step 10: Find the patient_id that appears most frequently

most_frequent_patient_id = patient_id_counts.idxmax()
most_frequent_count = patient_id_counts.max()

print("Most frequently appearing patient_id:", most_frequent_patient_id)
print("Number of times it appears:", most_frequent_count)
```

```python
# Step 11: Remove patients where patient_id appears more than two times

# Identify patient IDs that appear more than two times
patients_more_than_two = patient_id_counts[patient_id_counts > 2].index

# Filter the cleaned dataset to remove those patient IDs
cancer_df_filtered = cancer_df_clean[~cancer_df_clean['patient_id'].isin(patients_more_than_two)]

# Display shape before and after filtering
print("Shape before removing patients appearing more than two times:", cancer_df_clean.shape)
print("Shape after removing patients appearing more than two times:", cancer_df_filtered.shape)

# Display first 5 rows of filtered dataset
cancer_df_filtered.head()
```

```python
# Step 12: Create a new DataFrame containing patient_id and doctor_name

categorical_df = cancer_df_filtered[['patient_id', 'doctor_name']].copy()

# Add doctor_count column with value 1
categorical_df['doctor_count'] = 1

# Display the new categorical DataFrame
categorical_df.head()
```

```python
# Step 13: Pivot the DataFrame so each doctor's name becomes a column

doctor_pivot_df = categorical_df.pivot_table(
    index='patient_id',
    columns='doctor_name',
    values='doctor_count',
    aggfunc='sum',
    fill_value=0
)

# Display the pivoted DataFrame
doctor_pivot_df.head()
```

```python
# Step 14: Drop the multi-index from columns using droplevel()

# Check if the pivoted DataFrame has multi-index columns
if isinstance(doctor_pivot_df.columns, pd.MultiIndex):
    doctor_pivot_df.columns = doctor_pivot_df.columns.droplevel(0)
else:
    print("Columns are not MultiIndex, so droplevel() is not required for this output.")

# Reset index so patient_id becomes a normal column again
doctor_onehot_df = doctor_pivot_df.reset_index()

# Display the one-hot encoded doctor DataFrame
doctor_onehot_df.head()
```

```python
# Step 15: Display the one-hot encoded DataFrame

doctor_onehot_df
```

```python
# Step 16: Merge the one-hot encoded doctor DataFrame back to the filtered cancer dataset

combined_df = cancer_df_filtered.merge(
    doctor_onehot_df,
    on='patient_id',
    how='left'
)

# Display the final combined DataFrame
combined_df.head()
```

```python
# Step 17: Display the final combined DataFrame

combined_df
```

```python
# Step 18: Drop the doctor_name column from the combined DataFrame

combined_df = combined_df.drop(columns=['doctor_name'])

# Display the updated combined DataFrame
combined_df.head()
```

```python
# Step 19: Define a function to classify cell type as normal or abnormal

def celltypelabel(x):
    if ((x['cell_size_uniformity'] > 5) &
        (x['cell_shape_uniformity'] > 5)):
        return 'normal'
    else:
        return 'abnormal'
```

```python
# Step 20: Use apply() to create a new column named cell_type_label

combined_df['cell_type_label'] = combined_df.apply(celltypelabel, axis=1)

# Display the updated DataFrame
combined_df.head()
```

```python
# Final Step: Save the final combined DataFrame as a CSV file

combined_df.to_csv("lab7_final_combined_dataframe.csv", index=False)

print("Final combined DataFrame saved successfully as lab7_final_combined_dataframe.csv")
print("Final dataset shape:", combined_df.shape)
```
