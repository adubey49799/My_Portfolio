## Part 1: Working with Arrays

### Task 1
Create a NumPy array named `a` with shape `(3, 4)`.


````python
# 1. Create a NumPy array named a with shape (3, 4)
a = np.array([
    [3, 8, 15, 2],
    [2, 10, 5, 3],
    [4, 0, 2, 4]
])

print("Array a:")
print(a)
````

````python
# 3. Print the number of dimensions
print("Number of dimensions of array a:")
print(a.ndim)
````

Use the built-in `len()` function on `a`.


````python
# 4. Use built-in len() function
print("Result of len(a):")
print(len(a))
````

### Task 5
Based on the shape of the array, determine the number of dimensions and the length.


````python
# 5. Explanation based on shape
print("The shape of a is:", a.shape)
print("Because the shape has two values, the array has 2 dimensions.")
print("The first value in the shape is 3, so len(a) returns 3.")
print("This means len(a) gives the number of rows in this 2D array.")
````

## Part 3: Indexing and Slicing

Create another array named `a`.


````python
# 12. Create another array named a
a = np.array([
    [2, 7, 12, 0],
    [3, 9, 3, 4],
    [4, 0, 1, 3]
])

print("New array a:")
print(a)
````

## Part 4: Boolean Masks and Statistics

### Task 16
Recall array `a`.


````python
# 16. Recall array a
print("Original array a:")
print(a)
````

### Task 17
Use a comparison operation `> 5` to create a Boolean array.


````python
# 17. Create Boolean mask for values greater than 5
mask = a > 5

print("Boolean mask where elements are greater than 5:")
print(mask)
````

### Task 18
Use the mask to extract all values greater than 5.


````python
# 18. Extract all values greater than 5
values_greater_than_5 = a[mask]

print("Values greater than 5:")
print(values_greater_than_5)
````

### Task 19
Modify the array in-place so that all values greater than 5 are replaced with 5.


````python
# 19. Modify the array in-place so values greater than 5 become 5
a[a > 5] = 5

print("Modified array a after replacing values greater than 5 with 5:")
print(a)
````

## Part 5: Array Aggregation Functions

### Task 20
Using array `a` from earlier, compute the required aggregation values.

**Note:** Since Part 4 modifies `a` in-place, the aggregation below uses the modified version of `a`.


````python
print("Using modified array a from Part 4:")
print(a)
````

````python
# 20a. Sum of all values
total_sum = np.sum(a)

print("20a. Sum of all values:")
print(total_sum)
````

````python
# 20b. Sum of each column
column_sum = np.sum(a, axis=0)

print("20b. Sum of each column:")
print(column_sum)
````

````python
# 20c. Sum of each row
row_sum = np.sum(a, axis=1)

print("20c. Sum of each row:")
print(row_sum)
````

````python
# 20d. Mean of all values
mean_value = np.mean(a)

print("20d. Mean of all values:")
print(mean_value)
````

````python
# 20e. Minimum value
minimum_value = np.min(a)

print("20e. Minimum value:")
print(minimum_value)
````

````python
# 20f. Maximum value
maximum_value = np.max(a)

print("20f. Maximum value:")
print(maximum_value)
````
